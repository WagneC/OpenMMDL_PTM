# How to use  ProLif for PTM Reisdues from PROTLIG.pdb and LIG.sdf

```python
# Load files
LIGsdf = "/home/cwagner/openmmdl_ptm/OpenMMDL/openmmdl_ptm/maleimide.sdf"
PROTLIGpdb = "/home/cwagner/openmmdl_ptm/OpenMMDL/openmmdl_ptm/3ip9_dye.pdb"
```

```python
import prolif as plf
from rdkit import Chem

protein_file = str(plf.datafiles.PROTLIGpdb)

rdkit_prot = Chem.MolFromPDBFile(protein_file, removeHs=False)
protein_mol = plf.Molecule(rdkit_prot)
# histidine HSE347 not recognized by RDKit
plf.display_residues(protein_mol, slice(260, 263))
```

```python
import MDAnalysis as mda
import prolif as plf

u = mda.Universe(PROTLIGpdb)
protein_mol = plf.Molecule.from_mda(u)
plf.display_residues(protein_mol)
```
